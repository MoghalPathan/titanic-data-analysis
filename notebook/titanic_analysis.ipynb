Cleaning

6. Decide what to do with missing values in at least 2 columns — fill some (with mean/median/mode), drop others. .


```python
#fill missing age with median(robust to outliers), Embarked with the most common value

df["Age"] = df["Age"].fillna(df["Age"].median())   #age selects column, 'fillna' means replace missing values overall meaning -> finnd missing values in the Age column and fill them with meadian it means whenever age is missing, replace it with the median age
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

7. Find and remove duplicate rows, if any exist.

```python
# remove exact duplicate rows

before_duplication = len(df)   # len(df) tell us how many rows you currently have.
df = df.drop_duplicates()    #removes duplicate rows

print("Duplicates removed: ",before_duplication - len(df) )   # calculates how many rows were removed
```

9. Create a new column computed from existing ones (e.g. a family-size column from two other columns)

```python
# new column computed from existing ones

""" 
SibSp = siblings/spouses aboard
Parch = parents/children aboard
+1 is for the passenger themselves.
"""

df["Family size"] = df["SibSp"] + df["Parch"] + 1     # +1 to count the passenger themself
df.head()
```

Analysis

10. Group the data by one categorical column and compute the average of a numeric column per group (`groupby`).

```python
# average fare per class("What is the average fare for each passenger class?")
""" 
groupby("Pclass")
Groups passengers according to:
Pclass = 1
Pclass = 2
Pclass = 3

["Fare"] selects the Fare column.

.mean() alculates the average.

"""

avg_fair_per_class = df.groupby("Pclass")["Fare"].mean()
print(avg_fair_per_class)
```

```python
# top/bottom 5 by fare
# ascending=False means descending order.

top = df.sort_values("Fare", ascending=False).head()    #Sort passengers by Fare from highest to lowest, then show the first 5.
bottom = df.sort_values("Fare", ascending=True).head()  #Sort passengers by Fare from lowest to highest, then show the first 5.
```

```python
# distribution of a categorical column

print(df["Embarked"].value_counts())  #This counts how many times each value appears

# which means "How many passengers embarked from each location?"
```

```python
# filter with 2 combined conditions

filtered = df[(df["Pclass"] == 1) & (df["Sex"] == "female")]    # give me the passengers who are both first_class and female.
```

```python
# pivot table: average fare by class and sex


# What is the average fare for each passenger class, separated by sex?
pivot = pd.pivot_table(
    df,
    values="Fare",
    index="Pclass",
    columns="Sex",
    aggfunc="mean"
)
pivot
```
